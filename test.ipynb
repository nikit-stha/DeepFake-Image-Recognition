```python
L = torch.randn(2,4,5,6)

B, C, H, W = L.shape

y = F.adaptive_avg_pool2d(L, 1)
L_flat = L.view(B, C, -1)
y_flat = y.view(B, C, 1)

P = F.cosine_similarity(
    L_flat,
    y_flat,
    dim = 1
)

y.shape, L_flat.shape, y_flat.shape, P.shape
```
